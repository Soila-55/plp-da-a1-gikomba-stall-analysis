```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

## Task 1: Overall Sales Performance

I calculate the total, average, highest, and lowest daily sales for the 14 days. I also identify the day name and week number for the best and worst sales days.

```python
# Task 1: Overall Sales Performance

total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)

highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

highest_index = daily_sales.index(highest_sales)
lowest_index = daily_sales.index(lowest_sales)

highest_day = days[highest_index]
lowest_day = days[lowest_index

highest_week = (highest_index // 7) + 1
lowest_week = (lowest_index // 7) + 1

print("Total sales:", total_sales)
print("Average daily sales:", average_sales)
print("Highest daily sales:", highest_sales)
print("Highest sales day:", highest_day)
print("Highest sales week:", highest_week)
print("Lowest daily sales:", lowest_sales)
print("Lowest sales day:", lowest_day)
print("Lowest sales week:", lowest_week)
```

```python
# Task 2: Weekly Sales and Growth

def week_totals(sales):
    """Return the total sales for week 1 and week 2."""
    week1 = sum(sales[:7])
    week2 = sum(sales[7:14])
    return week1, week2


week1_total, week2_total = week_totals(daily_sales)

growth_percentage = ((week2_total - week1_total) / week1_total) * 100

print("Week 1 total:", week1_total)
print("Week 2 total:", week2_total)
print("Week-on-week growth:", round(growth_percentage, 2), "%")
```

```python
# Task 3: Classify Each Day

def classify_day(amount, average):
    """Classify a sales amount as Strong, Normal, or Weak based on the average."""
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    else:
        return "Normal"


day_categories = {}
category_counts = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}

for i in range(len(daily_sales)):
    category = classify_day(daily_sales[i], average_sales)
    day_categories[days[i] + " " + str(i + 1)] = category
    category_counts[category] += 1

print("Day classifications:")
for day, category in day_categories.items():
    print(day, ":", category)

print("\nCategory counts:")
print("Strong:", category_counts["Strong"])
print("Normal:", category_counts["Normal"])
print("Weak:", category_counts["Weak"])
```

```python
# Task 4: Weekday vs Weekend Performance

weekday_sales = []
weekend_sales = []

for i in range(len(daily_sales)):
    if days[i] in ["Mon", "Tue", "Wed", "Thu", "Fri"]:
        weekday_sales.append(daily_sales[i])
    else:
        weekend_sales.append(daily_sales[i])

weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)

percentage_difference = (
    (weekend_average - weekday_average) / weekday_average
) * 100

print("Weekday average sales:", round(weekday_average, 2))
print("Weekend average sales:", round(weekend_average, 2))
print("Weekend vs weekday difference:", round(percentage_difference, 2), "%")
```

```python
# Task 5: Stock Reorder and Gross Margin

def products_to_reorder(stock):
    """Return a list of products whose quantity is at or below the reorder level."""
    reorder_list = []

    for product, details in stock.items():
        if details["qty"] <= details["reorder"]:
            reorder_list.append(product)

    return reorder_list


reorder_products = products_to_reorder(stock)

print("Products to reorder:")
for product in reorder_products:
    print("-", product)


print("\nGross margin percentages:")

for product, details in stock.items():
    margin = ((details["price"] - details["cost"]) / details["price"]) * 100
    print(product + ":", round(margin, 2), "%")
```

```python
# Task 6: Stall Performance Report

print("========================================")
print("       STALL PERFORMANCE REPORT")
print("========================================")

print(f"Total sales: KES {total_sales:,}")
print(f"Average daily sales: KES {average_sales:,.2f}")
print(f"Highest daily sales: KES {highest_sales:,}")
print(f"Lowest daily sales: KES {lowest_sales:,}")

print(f"Week 1 total: KES {week1_total:,}")
print(f"Week 2 total: KES {week2_total:,}")
print(f"Week-on-week growth: {growth_percentage:.2f}%")

print(f"Weekday average: KES {weekday_average:,.2f}")
print(f"Weekend average: KES {weekend_average:,.2f}")
print(f"Weekend vs weekday difference: {percentage_difference:.2f}%")

print("Products to reorder:", ", ".join(reorder_products))

print("\nGross margins:")
for product, details in stock.items():
    margin = ((details["price"] - details["cost"]) / details["price"]) * 100
    print(f"{product}: {margin:.2f}%")

print("========================================")
```

## Recommendations for Wanjiku

1. Wanjiku should focus more on weekends because the average weekend sales of KES 4,625 were 54.68% higher than the weekday average of KES 2,990.

2. She should increase stock for Tomatoes and Sukuma because their quantities of 18 kg and 9 kg are at or below their reorder levels of 25 kg and 15 kg respectively.

3. Wanjiku should monitor the business growth closely because Week 2 sales of KES 26,350 were 19.50% higher than Week 1 sales of KES 22,050.

4. She should pay attention to products with higher gross margins, especially Avocado at 42.86% and Sukuma at 40.00%, while ensuring enough stock is available to meet customer demand.
